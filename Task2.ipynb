```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("Task 2 started successfully!")
```

```python
data = {
    "product_id": [101, 102, 103, 104, 104, 105, 106, 107],
    "product_name": [
        "Laptop", " Phone ", "Tablet", "Monitor",
        "Monitor", "Keyboard", None, "Mouse"
    ],
    "price": [
        "₹50,000", "25000", "₹18,500", None,
        None, "$100", "₹12,000", "1,500"
    ],
    "date": [
        "2026-01-10", "11/02/2026", "2026/03/15",
        "15-04-2026", "15-04-2026", "May 5, 2026",
        None, "2026-06-20"
    ],
    "category": [
        "Electronics", "electronics", "Electronics",
        "Display", "Display", "Accessories",
        "Accessories", None
    ]
}

df = pd.DataFrame(data)

display(df)
```

```python
# Remove duplicate records
df = df.drop_duplicates()

# Clean product names
df["product_name"] = df["product_name"].str.strip()

# Standardize category names
df["category"] = df["category"].str.title()

# Handle missing values
df["product_name"] = df["product_name"].fillna("Unknown")
df["category"] = df["category"].fillna("Uncategorized")

# Convert prices to numeric values
df["price"] = (
    df["price"]
    .str.replace("INR", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["price"] = pd.to_numeric(df["price"], errors="coerce")

# Convert dates to a consistent format
df["date"] = pd.to_datetime(df["date"], dayfirst=True, errors="coerce")

# Show cleaned dataset
display(df)
```

```python
required_columns = [
    "product_id",
    "product_name",
    "price",
    "date",
    "category"
]

print("Checking required columns...")

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

print("Missing columns:", missing_columns)
print("Duplicate rows:", df.duplicated().sum())
print("Missing values per column:")
print(df[required_columns].isnull().sum())

assert len(missing_columns) == 0, "Required columns are missing!"
assert df["product_id"].notna().all(), "Product IDs are missing!"
assert df["product_name"].notna().all(), "Product names are missing!"

print("Schema validation completed!")
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved!")
```

```python
report = f"""
DATHEON INTERNSHIP - TASK 2
Automated Data Cleaning & Schema Transformation

Final number of records: {len(df)}
Duplicate rows remaining: {df.duplicated().sum()}

Missing values per column:
{df.isnull().sum().to_string()}

Column data types:
{df.dtypes.to_string()}

Schema validation: Completed
"""

with open("validation_report.txt", "w") as file:
    file.write(report)

print(report)
```
